###Padronização de tipagem

```python
df_bruto.dtypes
```

```python
from numpy import int64
df_tipagem = df_bruto.copy()

df_tipagem["customerID"] = df_tipagem["customerID"].astype('string')
df_tipagem["CustomerFeedback"] = df_tipagem["CustomerFeedback"].astype('string')

df_tipagem['SeniorCitizen'] = df_tipagem['SeniorCitizen'].map({
    0: 'Não',
    1: 'Sim'
}) #p alterar o valor binario se eh idoso ou nao

#os tenure com duração explícita no feedback
correcoes_tenure = {
    1059: 72,
    1791: 12,
    1984: 72,
    2372: 1,
    2641: 36,
    2936: 15,
    3300: 6,
    3476: 4,
    3515: 6,
    4875: 28,
    5324: 24,
    5328: 72
}

# Índices sem duração exata no feedback
indices_sem_tenure_exato = [
    1097, 2361, 2896, 2959, 3153, 3388, 3837
]

# Transformar "unknown" em NA
df_tipagem["tenure"] = df_tipagem["tenure"].replace(
    r"(?i)^\s*unknown\s*$", pd.NA, regex=True
)

# Converter tenure para inteiro anulável
df_tipagem["tenure"] = pd.to_numeric(
    df_tipagem["tenure"], errors="coerce"
).astype("Int64")

# Aplicar as correções manuais
for indice, meses in correcoes_tenure.items():
    df_tipagem.loc[indice, "tenure"] = meses

# Manter os casos sem duração exata como ausentes
df_tipagem.loc[indices_sem_tenure_exato, "tenure"] = pd.NA


for coluna in ["MonthlyCharges", "TotalCharges", "MonthlyIncome"]:
    df_tipagem[coluna] = pd.to_numeric(
        df_tipagem[coluna]
    )

colunas_categoricas = [
    'gender',
    'SeniorCitizen',
    'Partner',
    'Dependents',
    'PhoneService',
    'MultipleLines',
    'InternetService',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies',
    'Contract',
    'PaperlessBilling',
    'PaymentMethod',
    'Churn'
]

for coluna in colunas_categoricas:
  if coluna in df_tipagem.columns:
    df_tipagem[coluna] =  df_tipagem[coluna].astype("category")

df_tipagem.info()
```
