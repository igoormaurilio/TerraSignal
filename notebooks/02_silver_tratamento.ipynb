## 4. Inspeção inicial dos dados

```python
def analisar_dataframe(df, nome_tabela):
    print(f"{"="*20} {nome_tabela.upper()} {"="*20}")
    print(f"Linhas: {df.shape[0]}\nColunas: {df.shape[1]}")
    display(f"Valores duplicados: {df.duplicated().sum()}")
    print(f"Valores nulos:")
    display(df.isna().sum())
    display(df.head(10))
    display(df.tail(10))
    display(df.sample(10))
    display(df.info())
```

### 5.1. Identificando os valores

### 5.2. Identificação manual dos valores `unknown` pela coluna **CostumerFeedback**

A coluna CostumerFeedback, as vezes indica quanto tempo o cliente está, ou esteve, vinculado a empresa.

```python
mascara = (
    df_bruto["tenure"]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("unknown")
)

for indice, linha in df_bruto.loc[mascara].iterrows():
    print(f"Índice: {indice}")
    print(f"Cliente: {linha['customerID']}")
    print(f"Feedback: {linha['CustomerFeedback']}")
    print("-" * 80)
```

```python
# duração explícita no feedback
correcoes_tenure = {
    1059: 72,
    1791: 12,
    1984: 72,
    2372: 1,
    2641: 36,
    2936: 15,
    3300: 6,
    3476: 4,
    3515: 6,
    4875: 28,
    5324: 24,
    5328: 72
}

# em que o feedback não permite saber o tempo exato
indices_sem_tenure_exato = [
    1097, 2361, 2896, 2959, 3153, 3388, 3837
]
```

### 5.3. Correção dos valores 

```python
df_valores_tratados = df_bruto.copy()
```

```python
# correções manuais para os valores conhecidos
for indice, meses in correcoes_tenure.items():
    df_valores_tratados.loc[indice, "tenure"] = str(meses) # str para evitar problemas de conversão, visto que a coluna original é do tipo str

# <NA> para os casos sem duração exata
df_valores_tratados.loc[indices_sem_tenure_exato, "tenure"] = pd.NA

# verificar
display(
    df_valores_tratados.loc[
        list(correcoes_tenure.keys()) + indices_sem_tenure_exato,
        ["customerID", "tenure"]
    ].sort_index()
)

print("Total de valores ausentes:", df_valores_tratados["tenure"].isna().sum())
```

## 7. Tratamento de valores ausentes

```python
df_valores_tratados.isna().sum()
```

### 7.1. Exclusão de linhas com valores ausentes na coluna **ternure**

```python
valores_ausentes_ternure = df_valores_tratados['tenure'].isna().sum()
total_valores = df_valores_tratados.shape[0]

porcentagem_ausentes_ternure = (valores_ausentes_ternure / total_valores) * 100

print(f"Valores ausentes na coluna tenure: {valores_ausentes_ternure} ({porcentagem_ausentes_ternure:.2f}%)")
```

Após a conversão e correção da coluna tenure, foram identificados registros com valores ausentes. Inicialmente, foi realizada uma investigação utilizando as informações disponíveis na coluna CustomerFeedback, permitindo recuperar manualmente parte dos valores. Entretanto, 7 registros permaneceram sem informações suficientes para determinar o tempo de permanência dos clientes. Essas 7 linhas representam apenas 0,12% do total de registros do DataFrame. Como essa variável é importante para as análises posteriores e não foi possível recuperar os valores com segurança, optou-se pela exclusão desses registros.

```python
df_valores_tratados = df_valores_tratados.dropna(subset=["tenure"])
```

```python
valores_ausentes_ternure = df_valores_tratados['tenure'].isna().sum()
total_valores = df_valores_tratados.shape[0]

porcentagem_ausentes_ternure = (valores_ausentes_ternure / total_valores) * 100

print(f"Valores ausentes na coluna tenure: {valores_ausentes_ternure} ({porcentagem_ausentes_ternure:.2f}%)")
```

### 7.2. Imputação de valores na coluna **TotalCharges**

```python
valores_ausentes_totalcharges = df_valores_tratados['TotalCharges'].isna().sum()
total_valores = df_valores_tratados.shape[0]

porcentagem_ausentes_totalcharges = (valores_ausentes_totalcharges / total_valores) * 100

print(f"Valores ausentes na coluna TotalCharges: {valores_ausentes_totalcharges} ({porcentagem_ausentes_totalcharges:.2f}%)")
```

Como a coluna tenure representa o tempo de permanência do cliente e MonthlyCharges indica sua cobrança mensal, foi possível estabelecer uma aproximação do valor total acumulado por meio da multiplicação dessas duas variáveis. Dessa forma, optou-se pela imputação em vez da exclusão dos registros, evitando uma redução desnecessária da base de dados.

```python
df_valores_tratados[df_valores_tratados['tenure'] == 0]
```

```python
# conversão das colunas para o tipo numérico
tenure = pd.to_numeric(
    df_valores_tratados["tenure"],
    errors="coerce"
)
monthly_charges = pd.to_numeric(
    df_valores_tratados["MonthlyCharges"],
    errors="coerce"
)

# identificação dos valores ausentes na coluna TotalCharges
valores_ausentes = df_valores_tratados["TotalCharges"].isna()

# inputação dos valores
df_valores_tratados.loc[valores_ausentes, "TotalCharges"] = (
    tenure.loc[valores_ausentes] *
    monthly_charges.loc[valores_ausentes]
)
```

```python
valores_ausentes_totalcharges = df_valores_tratados['TotalCharges'].isna().sum()
total_valores = df_valores_tratados.shape[0]

porcentagem_ausentes_totalcharges = (valores_ausentes_totalcharges / total_valores) * 100

print(f"Valores ausentes na coluna TotalCharges: {valores_ausentes_totalcharges} ({porcentagem_ausentes_totalcharges:.2f}%)")
```

### 9.2. Correção dos tipos de dados

```python
df_tipagem[colunas_categoricas] = df_tipagem[colunas_categoricas].astype("category")

df_tipagem[colunas_string] = df_tipagem[colunas_string].astype("string")

for coluna in colunas_numericas:
    df_tipagem[coluna] = pd.to_numeric(df_tipagem[coluna], errors="coerce")
```
